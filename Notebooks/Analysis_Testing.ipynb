1. mutation count:Total number of genetic mutations found in a tumor
Higher mutation count = more unstable tumor
-What we can analyze:
Does higher mutation count → worse survival?
Or does it respond better to treatment?
-Business Insight(BI):
“Patients with higher mutation counts may require more aggressive or targeted therapies.”

2. MSI MANTIS Score
Measures microsatellite instability (MSI) — a type of DNA repair failure
High MSI score → defective repair system
High MSI patients often respond very well to immunotherapy
-BI: “MSI-high patients represent a high-value group for immunotherapy treatments.”

3. TMB (Tumor Mutational Burden - nonsynonymous)
Number of mutations that actually change proteins
Not all mutations matter.
Some = harmless
Some = change protein → affect tumor behavior
-TMB counts the important ones
High TMB → more abnormal proteins
Makes tumor more visible to immune system
Often linked to better response to immunotherapy
BI:“High TMB patients are strong candidates for immune-based treatments.”

4. Buffa Hypoxia Score
Measures how low-oxygen (hypoxic) the tumor environment is
Tumors often grow in low oxygen areas
Low oxygen → more aggressive cancer
Harder to treat
High score = more hypoxia = worse conditions
Hypoxic tumors:Grow faster, Resist treatment
BI:“Patients with high hypoxia scores may show reduced treatment effectiveness and poorer survival outcomes.”

| Biomarker      | What it tells you   |
| -------------- | ------------------- |
| Mutation Count | Tumor instability   |
| MSI Score      | DNA repair failure  |
| TMB            | Important mutations |
| Hypoxia Score  | Tumor environment   |
